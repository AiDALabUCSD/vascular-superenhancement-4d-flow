```python
import pickle
from pathlib import Path

pkl_path = Path("/home/ayeluru/mnt/fourier/repository/vascular-superenhancement-4d-flow/cardiac_partition_key.pkl")

with open(pkl_path, "rb") as f:
    cardiac_partition_key = pickle.load(f)

cardiac_partition_key
```

```python
import pandas as pd

splits_df = pd.read_csv("../../splits/splits_01-15-26.csv")

pkl_training = set(cardiac_partition_key["training"])
split_lookup = dict(zip(splits_df["patient_id"], splits_df["split"]))

overlap = pkl_training & set(split_lookup.keys())

print(f"Pickle training set:    {len(pkl_training)} patients")
print(f"Splits CSV (all):       {len(split_lookup)} patients")
print(f"Overlap:                {len(overlap)} patients\n")

if overlap:
    for patient in sorted(overlap):
        print(f"  {patient:25s} -> {split_lookup[patient]}")
else:
    print("No overlap found.")
```
