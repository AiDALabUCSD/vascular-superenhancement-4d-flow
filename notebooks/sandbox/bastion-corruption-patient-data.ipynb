```python
import os
import time
import numpy as np
import nibabel as nib
from pathlib import Path

BASE = Path("/home/ayeluru/mnt/fourier/projects/vascular-superenhancement-4d-flow/all_patients/patient_data")

patients = [
    ("Tieriegi", "train"),
    ("Pemiggig", "train"),
    ("Ribieko", "train"),
    ("Cornuefor", "train"),
    ("Gudulung", "train"),
    ("Suquepog", "test"),
    ("Jacibkut", "train"),
    ("Tiepolem", "test"),
    ("Krujouque", "train"),
    ("Obdufar", "train"),
    ("Tupulou", "skip"),
    ("Tupotu", "train"),
    ("Kuquelok", "test"),
    ("Ebukhot", "skip"),
    ("Plofibe", "not in splits"),
    ("Suedrurnep", "validation"),
    ("Slathalu", "train"),
    ("Sithiroo", "train"),
    ("Upbapit", "validation"),
    ("Kihaquo", "train"),
]

def iter_niftis(root):
    for dp, _, fs in os.walk(root):
        for f in fs:
            if f.endswith((".nii", ".nii.gz")):
                yield Path(dp) / f

def force_read(path):
    img = nib.load(path)
    arr = np.asanyarray(img.dataobj)
    if arr.size:
        _ = float(arr.flat[0])
        _ = float(arr.flat[arr.size // 2])
        _ = float(arr.flat[-1])

results = []
bad_files = []

for idx, (patient, split) in enumerate(patients, start=1):
    root = BASE / patient
    print(f"\n[{idx}/{len(patients)}] {patient} ({split})")

    if not root.exists():
        print("  MISSING DIRECTORY")
        results.append((patient, split, 0, 0, "MISSING"))
        continue

    n_files = 0
    bad = 0
    t0 = time.time()

    for p in iter_niftis(root):
        n_files += 1
        try:
            force_read(p)
        except Exception as e:
            bad += 1
            bad_files.append((patient, p, str(e)))
            print("  BAD:", p)
            print("       ", e)

        if n_files % 50 == 0:
            print(f"  progress: {n_files} files | bad so far: {bad}")

    dt = time.time() - t0
    status = "OK" if bad == 0 else f"BAD={bad}"
    print(f"  DONE: {n_files} files | {status} | {dt:.1f}s")

    results.append((patient, split, n_files, bad, dt))

print("\n========== SUMMARY ==========")
for r in results:
    print(r)

print("\nTotal bad files:", sum(r[3] for r in results))
```
